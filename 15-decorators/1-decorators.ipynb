```python
def wrapper(func):
    def inner(*args, **kwargs):
        result = func(*args, **kwargs)
        return result
    return inner
```

```python
def add(a,b,c):
    return a + b + c

def greet(name):
    return f'Hello, {name}!'

def join(data,*,item_sep=',',line_sep='\n'):
    return line_sep.join(
        [
            item_sep.join(str(item) for item in row)
            for row in data
        ]
    )
```

```python
add(1,2,3)
```

```python
def log(func):
    def inner(*args, **kwargs):
        result = func(*args, **kwargs)
        print(f'{func.__name__} called... result={result}')
        return result
    return inner
```

```python
add_logged = log(add)
greet_logged = log(greet)
join_logged = log(join)
```

```python
add_logged(1,2,3)
```

```python
greet_logged('Python')
```

```python
from time import perf_counter
```

```python
def log(func):
    def inner(*args, **kwargs):
        start = perf_counter()
        result = func(*args, **kwargs)
        end = perf_counter()
        logger.debug(f'called={func.__name__} ,elapsed={end - start}')
        return result
    return inner
```

```python
@log
def add(a,b,c):
    return a + b + c

@log
def greet(name):
    return f'Hello, {name}!'

@log
def join(data,*,item_sep=',',line_sep='\n'):
    return line_sep.join(
        [
            item_sep.join(str(item) for item in row)
            for row in data
        ]
    )
```
