```python
def get_first_last(dt):
    start = datetime.date(year=dt.year,month=dt.month,day=1)

    if start.month == 12:
        new_year = start.year + 1
        new_month = 1
    else:
        new_year = dt.year
        new_month = dt.month + 1

    end = datetime.date(new_year,new_month,1) + datetime.timedelta(days=-1)

    return start,end
```
