```
def average(*values):
    try :
        return sum(values) / len(values)
    except ZeroDivisionError:
        return 0
```

```
average(0)
```

```
average(1,2,3)
```

```
average()
```

```
average(1,2,3)
```

```
def product(*values):
    prod = 1
    for v in values:
        prod *= v
    return prod
```

```
product(1,2,3)
```

```
product(1,2,3,4)
```

```
product()
```

```
def product(*values):
    if len(values) == 0:
        raise ValueError('Must provide at least one argument')
    prod = 1
    for v in values:
        prod *= v
    return prod
```

```
product(1,2,3,4)
```

```
product()
```

```
def product(*values):
    if len(values) == 0:
        raise ValueError('Must provide at least one argument')
    prod = 1
    for v in values:
        prod *= v
    return prod
```

```
l = [1,2,3,4]
```

```
product(l)
```

```
product(*l)
```
